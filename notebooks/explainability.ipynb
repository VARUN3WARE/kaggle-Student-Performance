# Explainability & Interpretability

This notebook reproduces the SHAP explainability plots and documents interpretations about what features most influence student performance predictions. It expects the processed dataset at `data/processed/processed_student_data.csv` and trained models under `models/`.

```python
# Imports
import pandas as pd
import joblib
import shap
import matplotlib.pyplot as plt

# Display settings
%matplotlib inline
```

```python
# Load processed data and models
df = pd.read_csv('../data/processed/processed_student_data.csv')
print('Data shape:', df.shape)
# Load a model for explainability (example: ridge)
model = joblib.load('../models/best_ridge_model.pkl')
# If a separate transformer exists, load it too
from pathlib import Path
if Path('../models/transformer.pkl').exists():
    transformer = joblib.load('../models/transformer.pkl')
else:
    transformer = None
```

```python
# Compute SHAP values (TreeExplainer recommended for tree models)
explainer = shap.Explainer(model)
# Prepare features: if transformer present, transform; otherwise pass df
try:
    X = transformer.transform(df) if transformer is not None else df
except Exception:
    X = df.values
shap_values = explainer(X)

# Summary plot
shap.summary_plot(shap_values, X)
```
